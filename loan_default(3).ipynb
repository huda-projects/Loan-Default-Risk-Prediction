## 1. Project Objective and Research Question

### Research Question

**Can we identify borrowers who are at higher risk of default so that a financial institution can make better lending and credit-risk decisions?**

### Project Objective

The objective of this project is to use historical borrower and loan data to investigate characteristics and patterns associated with loan default. ### Investigate Potential Outliers

```python
# Investigate Potential Outliers

# Identify numerical predictor variables.
numerical_predictors = df.select_dtypes(include=["int64", "float64"]).columns

# Exclude the target variable from the outlier analysis.
numerical_predictors = numerical_predictors[numerical_predictors != "Status"]

# Create a list to store the outlier results.
outlier_results = []

# Calculate IQR-based outlier thresholds for each numerical predictor.
for column in numerical_predictors:
    q1 = df[column].quantile(0.25)
    q3 = df[column].quantile(0.75)
    iqr = q3 - q1

    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    outlier_count = (
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ).sum()

    outlier_percentage = (outlier_count / len(df)) * 100

    outlier_results.append({
        "Feature": column,
        "Q1": q1,
        "Q3": q3,
        "IQR": iqr,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Outlier Count": outlier_count,
        "Outlier Percentage": outlier_percentage
    })

# Create a summary table.
outlier_summary = pd.DataFrame(outlier_results)

# Display the features with the largest number of potential outliers.
outlier_summary.sort_values("Outlier Count",ascending=False)
```

<span style="color:cadetblue;"><b></b>
The **term** feature shows a high outlier percentage (18.12%) because Q1 and Q3 are both 360, resulting in an IQR of zero. This indicates that most loans have a 360-month term, so we will examine its unique values to determine whether the flagged values are legitimate.</span>

## 5. Exploratory Data Analysis

Exploratory data analysis will be used to understand the characteristics of the dataset and identify patterns and relationships that may help answer the research question.

### Categorical Features and the Target Variable
This analysis examines the relationship between categorical borrower and loan features and the target variable, **Status**. The goal is to identify categories associated with higher or lower observed default rates and determine the size of the differences between categories.


```python
# Count and percentage of loan default status within each category.

categorical_features = df.select_dtypes(include="object").columns

results = []

for feature in categorical_features:

    summary = (
        df.groupby([feature, "Status"])
        .size()
        .unstack()
    )

    summary["Total"] = summary.sum(axis=1)

    summary["Status 0 %"] = summary[0] / summary["Total"] * 100
    summary["Status 1 %"] = summary[1] / summary["Total"] * 100

    summary = summary.reset_index()

    summary = summary.rename(columns={
        feature: "Category",
        0: "Status 0 Count",
        1: "Status 1 Count"
    })

    summary["Feature"] = feature

    results.append(summary)

categorical_status_summary = pd.concat(results, ignore_index=True)

categorical_status_summary = categorical_status_summary[
    [
        "Feature",
        "Category",
        "Total",
        "Status 0 Count",
        "Status 0 %",
        "Status 1 Count",
        "Status 1 %"
    ]
]

categorical_status_summary["Status 0 %"] = (
    categorical_status_summary["Status 0 %"].round(2)
)

categorical_status_summary["Status 1 %"] = (
    categorical_status_summary["Status 1 %"].round(2)
)

display(categorical_status_summary)
```

<span style="color:cadetblue;"><b></b>
To identify categorical features that may be associated with loan default, we compared the observed percentage of Status 1 (default) across the categories of each feature. We focused on features where the difference between the highest and lowest observed default rate was greater than approximately 10 percentage points. Based on this comparison, the five features selected for further analysis were **lump_sum_payment**, **Neg_ammortization**, **loan_type**, **business_or_commercial**, and **loan_purpose**. These features were then examined in greater detail through additional visual analysis to better understand their relationship with loan default and their potential usefulness for identifying higher-risk borrowers.</span>


```python
# Plot Status 0 % and Status 1 % for each category
# within each selected categorical feature.

import plotly.express as px



selected_features = [
    "lump_sum_payment",
    "Neg_ammortization",
    "loan_type",
    "business_or_commercial",
    "loan_purpose"
]

for feature in selected_features:

    plot_data = categorical_status_summary[
        categorical_status_summary["Feature"] == feature
    ].copy()

    plot_data = plot_data.melt(
        id_vars=["Category"],
        value_vars=["Status 0 %", "Status 1 %"],
        var_name="Status",
        value_name="Percentage"
    )

    fig = px.bar(
        plot_data,
        x="Category",
        y="Percentage",
        color="Status",
        barmode="group",
        text="Percentage",
        title=f"{feature}: Loan Default Status by Category",
        labels={
            "Category": "Category",
            "Percentage": "Percentage (%)",
            "Status": "Loan Status"
        },
        height=500
    )

    fig.update_traces(
        texttemplate="%{text:.1f}%",
        textposition="outside"
    )

    fig.update_yaxes(
        range=[0, 100],
        ticksuffix="%"
    )

    fig.update_layout(
        legend_title_text="Loan Status",
        margin=dict(
            l=70,
            r=50,
            t=90,
            b=80
        )
    )

    fig.show()
```

<span style="color:cadetblue;"><b></b>
For these five features, we observed a noticeable difference in the proportion of Status 0 (no default) and Status 1 (default) across their categories. This means that borrowers in different categories of these features show different observed default rates. From a business perspective, these features may help distinguish groups of borrowers with different levels of observed loan-default risk. These differences suggest that **lump_sum_payment**, **Neg_ammortization**, **loan_type**, **business_or_commercial**, and **loan_purpose** may be useful features for understanding and predicting loan default.</span>

### Numerical Features and the Target Variable

## 6. Preparing the Data for Modeling

After completing the cleaningb and exploratory analysis, the dataset will be prepared for machine learning.

```python
# Encode categorical features into 0/1 columns.

categorical_features = df.select_dtypes(include="object").columns.tolist()

df_encoded = pd.get_dummies(df,columns=categorical_features, drop_first=False)

# Convert True/False dummy values to 0/1.
dummy_columns = df_encoded.select_dtypes(include="bool").columns

df_encoded[dummy_columns] = df_encoded[dummy_columns].astype(int)

# Display the encoded data.
display(df_encoded.head())
```

```python
# Separate predictor features from the target and split the data.

from sklearn.model_selection import train_test_split

X = df_encoded.drop(columns=["Status"])
y = df_encoded["Status"]

X_train, X_test, y_train, y_test = train_test_split(X, y,test_size=0.20,random_state=42,stratify=y)

print("Training set:", X_train.shape)
print("Testing set:", X_test.shape)
```

```python
# Scale the numerical features.

from sklearn.preprocessing import StandardScaler

numerical_features = df.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

numerical_features.remove("Status")

scaler = StandardScaler()

X_train[numerical_features] = scaler.fit_transform(X_train[numerical_features])

X_test[numerical_features] = scaler.transform(X_test[numerical_features])

# Display the prepared training data.
display(X_train.head())
```

## 7. Baseline Classification Model

A baseline classification model will be developed to provide an initial benchmark for predicting loan default. The purpose of this model is to provide a measurable starting point against which additional models, improvements, and optimization techniques can be compared.
For Model evaluation and performance metric, the classification model will be evaluated using accuracy, precision, recall, and F1 score. Since the data is imbalanced, multiple metrics are used instead of relying on accuracy alone. A confusion matrix will also be used to examine correct and incorrect predictions and support comparison with other models and future improvements.

```python
# Build and evaluate the Logistic Regression model.

import time

from sklearn.linear_model import LogisticRegression

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    balanced_accuracy_score,
    confusion_matrix,
    ConfusionMatrixDisplay
)





logistic_model = LogisticRegression(
    class_weight="balanced",
    max_iter=1000
)

# Measure training time.
start_train = time.perf_counter()

logistic_model.fit(X_train, y_train)

training_time = time.perf_counter() - start_train

# Measure testing/prediction time.
start_test = time.perf_counter()

y_pred = logistic_model.predict(X_test)

testing_time = time.perf_counter() - start_test

# Calculate evaluation metrics.
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, zero_division=0)
recall = recall_score(y_test, y_pred, zero_division=0)
f1 = f1_score(y_test, y_pred, zero_division=0)

# Store results.
logistic_results = pd.DataFrame({
    "Model": ["Logistic Regression"],
    "Accuracy": [accuracy],
    "Precision": [precision],
    "Recall": [recall],
    "F1 Score": [f1],
    "Training Time (seconds)": [training_time],
    "Testing Time (seconds)": [testing_time]
})

display(logistic_results.round(4))
```

```python
from sklearn.model_selection import StratifiedKFold, GridSearchCV
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import  OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from imblearn.pipeline import Pipeline as ImbPipeline
from imblearn.under_sampling import RandomUnderSampler

from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import LinearSVC
```

```python
# Separate features (X) and target (y)
X_new = df.drop(columns=["Status"])
y_new = df["Status"]

# Split into training and testing sets (80/20)
X_train_new, X_test_new, y_train_new, y_test_new = train_test_split(
    X_new,
    y_new,
    test_size=0.20,
    random_state=42,
    stratify=y_new
)

# Identify numerical and categorical features
numerical_features = X_train_new.select_dtypes(
    include=["number"]
).columns.tolist()

categorical_features = X_train_new.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

# Configure 5-fold stratified cross-validation
cv = StratifiedKFold(
    n_splits=3,
    shuffle=True,
    random_state=42
)

# Store model results and best models
model_results = []
best_models = {}

# Display dataset information
print("Training set shape:", X_train_new.shape)
print("Testing set shape:", X_test_new.shape)

print("\nTraining target distribution:")
print(y_train_new.value_counts())

print("\nNumerical features:", len(numerical_features))
print("Categorical features:", len(categorical_features))
```

```python

# KNN: preprocessing, hyperparameter tuning, and evaluation

# Preprocess numerical and categorical features
numeric_transformer = StandardScaler()

categorical_transformer = OneHotEncoder(handle_unknown="ignore")

preprocessor = ColumnTransformer([
    ("num", numeric_transformer, numerical_features),
    ("cat", categorical_transformer, categorical_features)
])



knn_pipeline = ImbPipeline([
    ("preprocessor", preprocessor),
    ("sampler", RandomUnderSampler(random_state=42)),
    ("model", KNeighborsClassifier())
])

knn_param_grid = {
    "model__n_neighbors": [10, 15, 20],
    "model__weights": ["uniform", "distance"],
    "model__p": [1, 2]
}

knn_grid = GridSearchCV(
    estimator=knn_pipeline,
    param_grid=knn_param_grid,
    cv=cv,
    scoring="f1",
    n_jobs=-1,
    verbose=1
)

# Train and tune
start_time = time.time()
knn_grid.fit(X_train_new, y_train_new)
knn_train_time = time.time() - start_time

# Predict on untouched test data
start_time = time.time()
knn_pred = knn_grid.predict(X_test_new)
knn_prediction_time = time.time() - start_time

# Evaluate
knn_metrics = {
    "Model": "KNN",
    "Accuracy": accuracy_score(y_test_new, knn_pred),
    "Precision": precision_score(y_test_new, knn_pred, zero_division=0),
    "Recall": recall_score(y_test_new, knn_pred, zero_division=0),
    "F1": f1_score(y_test_new, knn_pred, zero_division=0),
    "Balanced Accuracy": balanced_accuracy_score(y_test_new, knn_pred),
    "Training Time (s)": knn_train_time,
    "Prediction Time (s)": knn_prediction_time,
    "Best CV F1": knn_grid.best_score_,
    "Best Parameters": knn_grid.best_params_
}

model_results.append(knn_metrics)
best_models["KNN"] = knn_grid.best_estimator_


print("\nTest-set evaluation:")
for metric, value in knn_metrics.items():
    print(f"{metric}: {value}")

ConfusionMatrixDisplay.from_predictions(
    y_test_new,
    knn_pred,
    display_labels=["Non-default", "Default"],
    cmap="Blues"
)
plt.title("KNN - Confusion Matrix")
plt.tight_layout()
plt.show()
```

```python

# Decision Tree: hyperparameter tuning and evaluation

tree_pipeline = ImbPipeline([
    ("preprocessor", preprocessor),
    ("model", DecisionTreeClassifier(random_state=42))
])

tree_param_grid = {
    "model__max_depth": [5, 10, 20, None],
    "model__min_samples_split": [2, 10, 20],
    "model__min_samples_leaf": [1, 5, 10],
    "model__criterion": ["gini", "entropy"],
    "model__class_weight": [None, "balanced"]
}

tree_grid = GridSearchCV(
    estimator=tree_pipeline,
    param_grid=tree_param_grid,
    cv=cv,
    scoring="f1",
    n_jobs=-1,
    verbose=1
)

start_time = time.time()
tree_grid.fit(X_train_new, y_train_new)
tree_train_time = time.time() - start_time

start_time = time.time()
tree_pred = tree_grid.predict(X_test_new)
tree_prediction_time = time.time() - start_time

tree_metrics = {
    "Model": "Decision Tree",
    "Accuracy": accuracy_score(y_test_new, tree_pred),
    "Precision": precision_score(y_test_new, tree_pred, zero_division=0),
    "Recall": recall_score(y_test_new, tree_pred, zero_division=0),
    "F1": f1_score(y_test_new, tree_pred, zero_division=0),
    "Balanced Accuracy": balanced_accuracy_score(y_test_new, tree_pred),
    "Training Time (s)": tree_train_time,
    "Prediction Time (s)": tree_prediction_time,
    "Best CV F1": tree_grid.best_score_,
    "Best Parameters": tree_grid.best_params_
}

model_results.append(tree_metrics)
best_models["Decision Tree"] = tree_grid.best_estimator_

print("Best parameters:", tree_grid.best_params_)
print("Best cross-validation F1:", round(tree_grid.best_score_, 4))
print("\nTest-set evaluation:")
for metric, value in tree_metrics.items():
    print(f"{metric}: {value}")

ConfusionMatrixDisplay.from_predictions(
    y_test_new,
    tree_pred,
    display_labels=["Non-default", "Default"],
    cmap="Blues"
)
plt.title("Decision Tree - Confusion Matrix")
plt.tight_layout()
plt.show()
```

```python

# Linear SVM: hyperparameter tuning and evaluation

svm_pipeline = ImbPipeline([
    ("preprocessor", preprocessor),
    ("model", LinearSVC(
        max_iter=5000,
        random_state=42
    ))
])

svm_param_grid = {
    "model__C": [0.01, 1, 5, 10],
    "model__class_weight": [None, "balanced"]
}

svm_grid = GridSearchCV(
    estimator=svm_pipeline,
    param_grid=svm_param_grid,
    cv=cv,
    scoring="f1",
    n_jobs=-1,
    verbose=1
)

start_time = time.time()
svm_grid.fit(X_train_new, y_train_new)
svm_train_time = time.time() - start_time

start_time = time.time()
svm_pred = svm_grid.predict(X_test_new)
svm_prediction_time = time.time() - start_time

svm_metrics = {
    "Model": "Linear SVM",
    "Accuracy": accuracy_score(y_test_new, svm_pred),
    "Precision": precision_score(y_test_new, svm_pred, zero_division=0),
    "Recall": recall_score(y_test_new, svm_pred, zero_division=0),
    "F1": f1_score(y_test_new, svm_pred, zero_division=0),
    "Balanced Accuracy": balanced_accuracy_score(y_test_new, svm_pred),
    "Training Time (s)": svm_train_time,
    "Prediction Time (s)": svm_prediction_time,
    "Best CV F1": svm_grid.best_score_,
    "Best Parameters": svm_grid.best_params_
}

model_results.append(svm_metrics)
best_models["Linear SVM"] = svm_grid.best_estimator_

print("Best parameters:", svm_grid.best_params_)
print("Best cross-validation F1:", round(svm_grid.best_score_, 4))
print("\nTest-set evaluation:")
for metric, value in svm_metrics.items():
    print(f"{metric}: {value}")

ConfusionMatrixDisplay.from_predictions(
    y_test_new,
    svm_pred,
    display_labels=["Non-default", "Default"],
    cmap="Blues"
)
plt.title("Linear SVM - Confusion Matrix")
plt.tight_layout()
plt.show()
```

```python
# Voting Classifier (Aggregation)
# Use best hyperparameters from previous individual models.
# Tune Logistic Regression within the Voting Classifier.

from sklearn.ensemble import VotingClassifier
from sklearn.metrics import classification_report

lr_pipe = ImbPipeline([
    ("preprocessor", preprocessor),
    ("model", LogisticRegression(
        class_weight="balanced",
        max_iter=1000,
        random_state=42
    ))
])

svm_pipe = ImbPipeline([
    ("preprocessor", preprocessor),
    ("model", LinearSVC(
        C=1,
        class_weight="balanced",
        max_iter=5000,
        random_state=42
    ))
])

dt_pipe = ImbPipeline([
    ("preprocessor", preprocessor),
    ("model", DecisionTreeClassifier(
        class_weight="balanced",
        criterion="gini",
        max_depth=10,
        min_samples_leaf=5,
        min_samples_split=20,
        random_state=42
    ))
])

knn_pipe = ImbPipeline([
    ("preprocessor", preprocessor),
    ("sampler", RandomUnderSampler(random_state=42)),
    ("model", KNeighborsClassifier(
        n_neighbors=20,
        p=2,
        weights="distance"
    ))
])

voting_model = VotingClassifier(
    estimators=[
        ("lr", lr_pipe),
        ("svm", svm_pipe),
        ("dt", dt_pipe),
        ("knn", knn_pipe)
    ],
    voting="hard",
    n_jobs=1
)

# Tune only Logistic Regression's C parameter
param_grid = {
    "lr__model__C": [0.01, 0.1, 1, 10, 100]
}

voting_grid = GridSearchCV(
    estimator=voting_model,
    param_grid=param_grid,
    scoring="f1",
    cv=cv,
    n_jobs=-1,
    verbose=1
)

# Training time
start = time.time()
voting_grid.fit(X_train_new, y_train_new)
voting_train_time = time.time() - start

best_voting = voting_grid.best_estimator_

# Prediction time
start = time.time()
voting_pred = best_voting.predict(X_test_new)
voting_prediction_time = time.time() - start

# Evaluation metrics
voting_results = {
    "Model": "Voting Classifier",
    "Accuracy": accuracy_score(y_test_new, voting_pred),
    "Precision": precision_score(y_test_new, voting_pred, zero_division=0),
    "Recall": recall_score(y_test_new, voting_pred, zero_division=0),
    "F1": f1_score(y_test_new, voting_pred, zero_division=0),
    "Balanced Accuracy": balanced_accuracy_score(y_test_new, voting_pred),
    "Training Time (s)": voting_train_time,
    "Prediction Time (s)": voting_prediction_time,
    "Best CV F1": voting_grid.best_score_,
    "Best Parameters": voting_grid.best_params_
}

model_results.append(voting_results)
best_models["Voting Classifier"] = best_voting

for metric, value in voting_results.items():
    print(f"{metric}: {value}")

print("\nClassification Report:")
print(classification_report(y_test_new, voting_pred, zero_division=0))

# Confusion matrix
cm = confusion_matrix(y_test_new, voting_pred)
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues")
plt.title("Voting Classifier - Confusion Matrix")
plt.xlabel("Predicted Label")
plt.ylabel("Actual Label")
plt.tight_layout()
plt.show()
```

```python
# Deep Neural Network (DNN)

import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from sklearn.base import clone


# 1. Set random seeds for reproducibility
np.random.seed(42)
tf.random.set_seed(42)

nn_preprocessor = ColumnTransformer([
    ("num", StandardScaler(), numerical_features),
    ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_features)
])

X_train_nn = nn_preprocessor.fit_transform(X_train_new)
X_test_nn = nn_preprocessor.transform(X_test_new)

# Convert data into the format expected by Keras
if hasattr(X_train_nn, "toarray"):
    X_train_nn = X_train_nn.toarray()

if hasattr(X_test_nn, "toarray"):
    X_test_nn = X_test_nn.toarray()

X_train_nn = np.asarray(X_train_nn, dtype=np.float32)
X_test_nn = np.asarray(X_test_nn, dtype=np.float32)

y_train_nn = np.asarray(y_train_new, dtype=np.float32)
y_test_nn = np.asarray(y_test_new, dtype=int)

# Build the Deep Neural Network
dnn = keras.Sequential([
    keras.Input(shape=(X_train_nn.shape[1],)),
    layers.Dense(16, activation="relu"),
     layers.Dense(8, activation="relu"),
     layers.Dense(32, activation="relu"),
    layers.Dense(16, activation="relu"),
    layers.Dropout(0.2),
    layers.Dense(1, activation="sigmoid")
])

# Configure the model
dnn.compile(
    optimizer="rmsprop",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

# Train the model
start = time.time()

history = dnn.fit(
    X_train_nn,
    y_train_nn,
    epochs=40,
    batch_size=128,
    validation_split=0.15,
    class_weight={0: 1.0, 1: 3.0},
    verbose=1
)

dnn_train_time = time.time() - start

# Make predictions
start = time.time()

dnn_prob = dnn.predict(X_test_nn, verbose=0).ravel()
dnn_pred = (dnn_prob >= 0.5).astype(int)

dnn_prediction_time = time.time() - start

# Evaluate the model
dnn_results = {
    "Model": "Deep Neural Network",
    "Accuracy": accuracy_score(y_test_nn, dnn_pred),
    "Precision": precision_score(
        y_test_nn, dnn_pred, zero_division=0
    ),
    "Recall": recall_score(
        y_test_nn, dnn_pred, zero_division=0
    ),
    "F1": f1_score(
        y_test_nn, dnn_pred, zero_division=0
    ),
    "Balanced Accuracy": balanced_accuracy_score(
        y_test_nn, dnn_pred
    ),
    "Training Time (s)": dnn_train_time,
    "Prediction Time (s)": dnn_prediction_time,

}

model_results.append(dnn_results)
best_models["Deep Neural Network"] = dnn

print("DNN Results:")
print(dnn_results)
```

After testing different DNN configurations by varying the number of hidden layers, neurons, dropout rate, batch size, and number of epochs, we selected the configuration with 4 hidden layers, a dropout rate of 0.2, a batch size of 128, and 40 epochs. We also applied class weights to improve the detection of loan defaults. We selected it because it provided a good balance between precision and recall while meeting our goal of keeping recall above 50%.
